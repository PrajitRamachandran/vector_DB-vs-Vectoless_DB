```python
import getpass
from openai import OpenAI

# 1. Securely enter your Mistral API key
api_key = getpass.getpass("Enter Mistral API Key: ")

client = OpenAI(
    base_url="https://api.mistral.ai/v1",
    api_key=api_key
)

# 2. Fetch and display all active Mistral model names
print("--- Available Mistral Models ---")
try:
    models = client.models.list()
    model_ids = [m.id for m in models.data]
    
    for model_id in model_ids:
        print(f"• {model_id}")
except Exception as e:
    print(f"Failed to fetch models. Check your API key. Error: {e}")

# 3. Fetch live rate limits using a minimal dummy request
print("\n--- Live Account Limits ---")
try:
    # We will ping 'mistral-small-latest' as it is extremely fast and cheap
    target_model = "mistral-small-latest" if "mistral-small-latest" in model_ids else model_ids[0]
    
    raw_response = client.chat.completions.with_raw_response.create(
        model=target_model,
        messages=[{"role": "user", "content": "ping"}],
        max_tokens=1
    )
    
    headers = raw_response.headers
    
    # Mistral uses slightly different header names than Cerebras
    print(f"Requests Remaining (Minute): {headers.get('x-ratelimit-remaining-req-minute', 'N/A')}")
    print(f"Tokens Remaining (Minute):   {headers.get('x-ratelimit-remaining-tokens-minute', 'N/A')}")
    print(f"Tokens Remaining (Month):    {headers.get('x-ratelimit-remaining-tokens-month', 'N/A')}")
    
except Exception as e:
    print(f"Could not fetch limits. Error: {e}")
```
